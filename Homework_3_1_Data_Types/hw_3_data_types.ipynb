## 1. Рядки (Strings)

```python
def string_length(string: str) -> int:
    """Return the length of a string."""
    return len(string)


string_length("I love Python")
```

```python
def string_concatenation(first_string: str, second_string: str) -> str:
    """Concatenate two strings."""
    return first_string + " " + second_string


string_concatenation("I love Python", "and Machine Learning")
```

## 2. Числа (Int/float)

```python
def square_of_a_number(number: float) -> int | float:
    """Return the square of a number."""
    return number**2


square_of_a_number(25)
```

```python
def sum_of_numbers(first_number: float, second_number: float) -> int | float:
    """Return the sum of two numbers."""
    return first_number + second_number


sum_of_numbers(25, 45)
```

```python
def division(dividend: int, divisor: int) -> tuple[int, int]:
    """Return the whole part and remainder of a division."""
    return dividend // divisor, dividend % divisor


division(21, 5)
```

## 3. Списки (Lists)

```python
def average(numbers: list[int | float]) -> float:
    """Return the average of a list of numbers."""
    return sum(numbers) / len(numbers)


average([1, 2, 3, 4, 5, 6, 7, 8])
```

```python
def find_common_elements(
    first_list: list[int | float], second_list: list[int | float]
) -> list[int | float]:
    """Return elements that are present in both lists."""
    return list(set(first_list).intersection(second_list))


find_common_elements([1, 2, 3, 4, 5, 5], [4, 5, 6, 7, 8])
```

## 4. Словники (Dictionaries)

```python
def print_dictionary_keys(dictionary: dict) -> None:
    """Print all keys of a dictionary."""
    print(*dictionary.keys(), sep="\n")


print_dictionary_keys({"name": "Sasha", "age": 18, "city": "Lviv"})
```

```python
def dict_union(first_dictionary: dict, second_dictionary: dict) -> dict:
    """Return a new dictionary containing both dictionaries."""
    return first_dictionary | second_dictionary


first_dict: dict = {"name": "Sasha", "age": 18}

second_dict: dict = {"city": "Lviv", "language": "Python"}

dict_union(first_dict, second_dict)
```

## 5. Множини (Sets)

```python
def set_union(
    first_set_value: set[int | float], second_set_value: set[int | float]
) -> set[int | float]:
    """Return the union of two sets."""
    return first_set_value.union(second_set_value)


first_set: set[int | float] = {1, 2, 3, 4}
second_set: set[int | float] = {3, 4, 5, 6}

set_union(first_set, second_set)
```

```python
def is_subset(
    first_set_value: set[int | float], second_set_value: set[int | float]
) -> bool:
    """Check whether the first set is a subset of the second set."""
    return first_set_value.issubset(second_set_value)


first_set: set[int | float] = {1, 2, 3, 4}
second_set: set[int | float] = {1, 2, 3, 4, 5, 6}

is_subset(first_set, second_set)
```

## 6. Умовні вирази та цикли

```python
def check_number_parity(number: int) -> str:
    """Return whether a number is even or odd."""
    if number % 2 == 0:
        return "Парне"
    return "Непарне"


print(check_number_parity(6))
print(check_number_parity(5))
```

```python
def only_even_from_list(numbers: list[int]) -> list[int]:
    """Return a list containing only even numbers."""
    even_numbers = []

    for number in numbers:
        if number % 2 == 0:
            even_numbers.append(number)

    return even_numbers


only_even_from_list([2, 6, 3, 4, 5, 7, 3, 7, 3, 5, 7, 8, 9, 10, 14, 4, 6, 8])
```

## 7. Написати лямбда-функцію визначальну парне/непарне.

```python
check_even_odd = lambda number: "Парне" if number % 2 == 0 else "Непарне"

print(check_even_odd(4))
print(check_even_odd(5))
```
