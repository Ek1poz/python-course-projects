## 1. Рядки (Strings)

```python
def string_length(string: str) -> int:
    """Return the length of a string."""
    return len(string)

string_length("I love Python")
```

```python
def string_concatenation(string1: str, string2: str) -> str:
    """Concatenate two strings."""
    return string1 + " " + string2

string_concatenation("I love Python", "and Machine Learning")
```

## 2. Числа (Int/float)

```python
def square_of_a_number(number: int | float) -> int | float:
    """Return the square of a number."""
    return number**2

square_of_a_number(25)
```

```python
def sum_of_numbers(num1: int | float, num2: int | float) -> int | float:
    """Return the sum of two numbers."""
    return num1 + num2

sum_of_numbers(25, 45)
```

```python
def division(num1: int, num2: int) -> tuple[int, int]:
    """Return the whole part and remainder of a division."""
    return num1 // num2, num1 % num2

division(21, 5)
```

## 3. Списки (Lists)

```python
def average(numbers: list[int | float]) -> float:
    """Return the average of a list of numbers."""
    return sum(numbers) / len(numbers)

average([1, 2, 3, 4, 5, 6, 7, 8])
```

```python
def mutual_elements(
    list1: list[int | float], 
    list2: list[int | float]
) -> list[int | float]:
    """Return elements that are present in both lists."""
    return list(set(list1).intersection(list2))

mutual_elements([1, 2, 3, 4, 5, 5], [4, 5, 6, 7, 8])
```

## 4. Словники (Dictionaries)

```python
def keys_from_dict(data: dict) -> None:
    """Print all keys of a dictionary."""
    print(*data.keys(), sep="\n")

keys_from_dict({
    "name": "Sasha",
    "age": 18,
    "city": "Lviv"
})
```

```python
def dict_union(data1: dict, data2: dict) -> dict:
    """Return a new dictionary containing both dictionaries."""
    return data1 | data2

dict1 = {
    "name": "Sasha",
    "age": 18
}

dict2 = {
    "city": "Lviv",
    "language": "Python"
}

dict_union(dict1, dict2)
```

## 5. Множини (Sets)

```python
def set_union(
    set1: set[int | float], 
    set2: set[int | float]
) -> set[int | float]:
    """Return the union of two sets."""
    return set1.union(set2)

set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

set_union(set1, set2)
```

```python
def is_subset(
    set1: set[int | float], 
    set2: set[int | float]
) -> bool:
    """Check whether the first set is a subset of the second set."""
    return set1.issubset(set2)

set1 = {1, 2, 3, 4}
set2 = {1, 2, 3, 4, 5, 6}

is_subset(set1, set2)
```

## 6. Умовні вирази та цикли

```python
def odd_even(num: int) -> str:
    """Return whether a number is even or odd."""
    if num % 2 == 0:
        return "Парне"
    return "Непарне"

print(odd_even(6))
print(odd_even(5))
```

```python
def only_even_from_list(numbers: list[int]) -> list[int]:
    """Return a list containing only even numbers."""
    only_even = []
    
    for i in numbers:
        if i % 2 == 0:
            only_even.append(i)

    return only_even

only_even_from_list([2, 6, 3, 4, 5, 7, 3, 7, 3, 5, 7, 8, 9, 10, 14, 4, 6, 8])
```

## 7. Написати лямбда-функцію визначальну парне/непарне.

```python
check_even_odd = lambda num: "Парне" if num % 2 == 0 else "Не парне"

print(check_even_odd(4))
print(check_even_odd(5))
```
